## This script: Prepare the moral part of the Morality-Conditioned PCT prompt

### Benchmark: Moral Foundation Questionnair - include five moral foundations

```python
from morality_prompt_utils import all_morality_cases, morality_groups, morality_options, morality_topics
import pandas as pd
import itertools

def build_morality_grid_csv(output_path):
    cases = all_morality_cases()
    groups = morality_groups()
    topics_order = morality_topics()

    score_map = {0: -3, 1: -2, 2: -1, 3: 1, 4: 2, 5: 3}

    rows = []
    for topic in topics_order:
        topic_cases = groups[topic]  # dict: id -> question
        
        q_ids = sorted(topic_cases.keys())
        q_texts = [topic_cases[qid] for qid in q_ids]

        
        options_text_for_each_q = []
        for qid in q_ids:
            key = "Relevance" if 1 <= qid <= 16 else "Agreement"
            opts = morality_options(key)
            
            options_text_for_each_q.append([opts[i] for i in range(6)])

        for combo in itertools.product(range(6), repeat=6):
            
            scores = [score_map[i] for i in combo]
            avg_score = sum(scores) / 6.0

            
            row = {
                "topic": topic,
                "question id": ",".join(str(i) for i in q_ids),
            }
            
            for idx in range(6):
                q_col = f"question {idx+1}"
                a_col = f"answer {idx+1}"
                row[q_col] = q_texts[idx]
                
                row[a_col] = f"{combo[idx]} sep|sep {options_text_for_each_q[idx][combo[idx]]}"

            row["score"] = avg_score
            rows.append(row)

    df = pd.DataFrame(rows)
    df.to_csv(output_path, index=False)


    # ===== statistics =====
    print(f"Rows: {len(df):,}")
    print(f"Columns: {len(df.columns)}")

    topic_counts = df["topic"].value_counts().sort_index()
    print("\nRows of each topic:")
    for t, n in topic_counts.items():
        print(f"  {t:<25} {n:,}")

    print("\nScore statistics:")
    print(df["score"].describe().to_string(float_format=lambda x: f"{x:,.3f}"))

    score_freq = df["score"].value_counts().sort_index()
    print("\nScore frequency distribution (top 10 items):")
    print(score_freq.head(10).to_string())
    
    return df

if __name__ == "__main__":
    build_morality_grid_csv(output_path="./politic_morality/morality/data/morality_prompt_raw.csv")
```
